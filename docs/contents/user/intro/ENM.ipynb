# Elastic network models

An elastic network represents selected molecular positions as nodes connected
within a cutoff. ElastNetMT uses normalized unit springs.

- **GNM** uses a scalar Kirchhoff matrix and describes isotropic fluctuations.
  A connected graph has one rigid zero mode.
- **ANM** uses a directional Hessian with three coordinates per node. A fully
  constrained network has six rigid zero modes; connectivity alone is not
  enough to establish rigidity.

Coordinates must be finite and distinct. GNM requires at least two nodes and
ANM three. Selections use MolSysMT syntax; node order follows the model's
`atom_indices` in its converted molecular system.

## Units and interpretation

Eigenvalues, eigenvectors and raw GNM fluctuations are dimensionless.
Experimental B-factor fitting produces magnitudes in square angstroms.
ANM trajectories apply an explicit physical amplitude after interpolation.
Physical spring stiffness and physical mode frequencies are not implemented.

## Engines and failures

`vectorized` uses NumPy; `parallel` requires Numba; `gpu` requires CuPy.
The current core `auto` route selects Numba when discoverable, otherwise
NumPy. Explicit engine failures propagate. Trajectory interpolation has its
own engine argument and defaults to `vectorized`.

Catch the documented [scientific errors](../../../api/developers/exceptions/api_exceptions.rst)
to revise inputs. Failed calibration or cutoff optimization preserves the
previous model state. See the [models](../models/index.md) for runnable examples.
