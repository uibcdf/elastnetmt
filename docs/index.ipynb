# ElastNetMT

Elastic Networks MultiToolkit provides Gaussian and anisotropic network models
for molecular structures. It uses MolSysMT for molecular data and PyUnitWizard
for explicit unit conversion.

```{image} _static/logo.svg
:width: 45%
:alt: ElastNetMT logo
```

Start with the [executable quickstart](contents/showcase/Quickstart.ipynb),
the [user guide](contents/user/index.md), or the [API reference](api/index.md).

Current source targets Python 3.11–3.14. Public installed delivery is being
qualified; follow the [installation status](contents/about/installation.md)
before choosing an installation route. The examples below use compatible
installed dependencies and locally bundled molecular data.

```{toctree}
:maxdepth: 2
:hidden:

contents/about/index
contents/showcase/index
contents/user/index
contents/developer/index
api/index
```
